```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(self.name, self.marks)

s1 = Student("Rahul", 85)
s2 = Student("Anita", 90)

s1.display()
s2.display()
```

```python
class Laptop:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

l1 = Laptop("HP", 50000)
print(l1.brand, l1.price)
```

```python
class Rectangle:
    def __init__(self, l, w):
        self.l = l
        self.w = w

    def area(self):
        return self.l * self.w

r = Rectangle(5, 4)
print(r.area())
```

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

class Manager(Employee):
    def __init__(self, name, salary, dept):
        super().__init__(name, salary)
        self.dept = dept

m = Manager("Raj", 50000, "IT")
print(m.name, m.salary, m.dept)
```

```python
class Father:
    def showFather(self):
        print("Father")

class Mother:
    def showMother(self):
        print("Mother")

class Child(Father, Mother):
    pass

c = Child()
c.showFather()
c.showMother()
```

```python
class BankAccount:
    def withdraw(self, amount):
        print("Withdraw:", amount)

class SavingsAccount(BankAccount):
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if self.balance - amount < 1000:
            print("Minimum balance required")
        else:
            self.balance -= amount
            print("Withdrawn:", amount)

s = SavingsAccount(5000)
s.withdraw(4500)
```

```python
class Account:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amt):
        self.__balance += amt

    def withdraw(self, amt):
        self.__balance -= amt

    def show(self):
        print(self.__balance)

a = Account()
a.deposit(1000)
a.withdraw(200)
a.show()
```

```python
class Circle:
    def area(self):
        return 3.14 * 5 * 5

class Square:
    def area(self):
        return 4 * 4

shapes = [Circle(), Square()]

for s in shapes:
    print(s.area())
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Bike(Vehicle):
    def start(self):
        print("Bike starts")

class Car(Vehicle):
    def start(self):
        print("Car starts")

b = Bike()
b.start()
```

```python
class InsufficientBalanceError(Exception):
    pass

balance = 1000

try:
    amt = int(input("Enter withdrawal amount: "))
    
    if amt > balance:
        raise InsufficientBalanceError("Not enough balance")
    else:
        balance -= amt
        print("Transaction successful")
        print("Remaining balance:", balance)

except InsufficientBalanceError as e:
    print(e)

except ValueError:
    print("Invalid input")
```
